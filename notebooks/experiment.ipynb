```python
train = []
val = []

with open(
    "../data/processed/splits/"
    "a9a67c231903bc2e3e71afead1d6daaf6b7201211fc25a05efd2bb2ee2d5e898/"
    "prepared-records.jsonl"
) as f:
    for line in f:
        record = json.loads(line)
        if (
            record["corpus"] == "sentfin"
            and len(record["native_record"]["sentiments"]) == 1
        ):
            if record["partition"] == "train":
                train.append(record)
            if record["partition"] == "validation":
                val.append(record)

print(len(train))
print(len(val))
```
